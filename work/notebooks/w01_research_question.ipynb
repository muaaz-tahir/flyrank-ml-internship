## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

Unit of analysis: one row = one content page (content_id).

Decision this improves: which pages the content review team looks at first,
out of everything in the inventory, given limited review capacity.

Action someone takes: a reviewer opens the top-ranked pages and decides
whether to refresh, expand, protect, prune, or simply monitor each one.

Cost of a wrong recommendation: if a page is ranked high but did not
actually need attention, the team loses reviewer time on a low-value page,
and a page that did need attention gets looked at later than it should.
This is a low-to-moderate cost mistake, since no page is changed
automatically. A human always makes the final call.

Why data or ML can help at all: with tens of thousands of pages and limited
review capacity, someone has to decide an order. A simple fixed rule (like
"page is old and still gets traffic") is a reasonable start, but it misses
pages where several weaker signals combine into real risk. A model can
combine signals like age, traffic trend, position, and word count into one
ranked list, which a fixed rule alone cannot do as well.

## 3. Quick look at the data (2-3 real numbers)

```python
!git clone https://github.com/muaaz-tahir/flyrank-ml-internship.git
import pandas as pd

df = pd.read_csv("flyrank-ml-internship/data/raw/content_refresh_anonymized.csv")

print(f"Total rows: {len(df)}")

# Number 1: how many pages are stale but still getting traffic
stale_visible = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]
print(f"Stale but visible pages: {len(stale_visible)} ({len(stale_visible)/len(df)*100:.1f}%)")

# Number 2: how many pages are declining with real demand behind them
declining_with_demand = df[(df["trend_direction"] == "down") & (df["impressions_90d"] >= 100)]
print(f"Declining pages with demand: {len(declining_with_demand)} ({len(declining_with_demand)/len(df)*100:.1f}%)")

# Number 3: overall trend split
print(df["trend_direction"].value_counts(normalize=True) * 100)
```
